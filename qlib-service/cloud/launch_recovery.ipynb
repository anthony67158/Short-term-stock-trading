# 量化基线恢复启动器

只启动实验环境恢复脚本，不访问生产 OSS。Token 仅保存在子进程内存中。

```python
import getpass
import os
import pathlib
import subprocess

root = pathlib.Path('/mnt/workspace/quant-lab')
log_dir = root / 'full477'
log_dir.mkdir(exist_ok=True)
token = getpass.getpass('Tushare Token: ')
env = os.environ.copy()
env['TUSHARE_TOKEN'] = token
env['TUSHARE_URL'] = 'https://ts.gyzcloud.top/api'
log_path = log_dir / 'notebook_launcher.log'
log_handle = open(log_path, 'ab', buffering=0)
process = subprocess.Popen(
    ['bash', str(root / 'r')],
    cwd=root,
    env=env,
    stdout=log_handle,
    stderr=subprocess.STDOUT,
    start_new_session=True,
)
token = None
env.pop('TUSHARE_TOKEN', None)
print('RECOVERY_PID', process.pid)
print('LOG_PATH', log_path)
```
